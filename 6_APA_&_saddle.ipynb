```python
import hicona
import cooler
import pandas as pd
import numpy as np
import cooltools
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import pybedtools
import os
```

### 1) APA

```python
DF=[]
flank=500000
resolution=25000
path='/shares/CIBIO-Storage/CIBIO/sharedCE/Leonardo_Morelli/HiCONA/articolo/cell_cycle_Dekker/cools/'

bins2=pd.read_csv('bins/bins_25000.csv')
bins2=bins2[[x for x in bins2.columns if x.startswith('G1')]+[x for x in bins2.columns if x.startswith('prometa')]]
files=os.listdir(path)

for cool in files[::-1]:
    CONDS=[]
    PHS=[]
    PHST=[]
    CLS=[]
    SCORES=[]
    MARK=[]
    print('cool')
    condition=cool.split('_')[1]
    PHASE=cool.split('_')[2]
    clr=cooler.Cooler(f'{path}{cool}::/resolutions/{resolution}')
    # Compute expected
    expected = cooltools.expected_cis(
        clr,
        nproc=8
    )
    bins=clr.bins()[:]
    bins=pd.concat([bins.T,bins2.T]).T
    if condition.startswith('ran'):
        anno=cooler.annotate(pix_ran,bins)

    else:
        anno=cooler.annotate(pix_del,bins)
    
    anno['code1']=anno['chrom1']+'_'+anno['start1'].astype(str)+'_'+anno['end1'].astype(str)
    anno['code2']=anno['chrom2']+'_'+anno['start2'].astype(str)+'_'+anno['end2'].astype(str)
    anno['pair2']=anno['code1']+'-'+anno['code2']
    anno['pair2']=['-'.join(sorted(x.split('-'))) for x in anno['pair2']]


    
    for cluster in sorted(list(set(anno['cluster']))):
        for mark in ['H3K27ac']:
            if condition.startswith('ran'):
                if mark=='ATAC':
                    phases=['prometaran','G1ran']
                else:
                    phases=['prometa','G1con']
            else:
                if mark=='ATAC':
                    phases=['prometadel','G1del']
                else:
                    phases=['prometa','G1del']

            for phase in phases:
                annof=anno[(anno['cluster']==cluster)&((anno[f'{phase}_{mark}1']!=0)&(anno[f'{phase}_{mark}2']!=0))]
                if annof.shape[0]==0:
                    print(f' not enough {mark} in cluster {cluster}')
                    continue
                else:
                    
                    annof=annof.drop_duplicates('pair')
                    loops=annof[['chrom1','start1','end1','chrom2','start2','end2']]
                    for c in loops.columns:
                        if c.startswith('chr'):
                            loops[c]=loops[c].astype(str)
                        else:
                            loops[c]=loops[c].astype(int)
    
    
                    # Generate aggregate pileup
                    print('aggregating pileups')
                    
                    stack = cooltools.pileup(
                        clr,
                        loops,
                        view_df=None,
                        expected_df=expected,
                        flank=flank,   # +/- 100 kb around anchors
                        nproc=8
                    )
                    print('computing APA')
                    
                    
                    apa = np.nanmean(stack, axis=0)
                    sns.heatmap(
                        apa,
                        vmax = 2.5,
                        center=1,
                        vmin = -.5,
                        cmap='coolwarm')
                    n = apa.shape[0]

                    #NEW CENTER --> -3:center:+3
                    
                    center = apa[(n//2)-3:(n//2)+4, (n//2)-3:(n//2)+4]
                    peak = np.nanmean(center)
                    #classic apa

                    

                    #dounut APA
                    # donut background
                    yy, xx = np.indices(apa.shape)
                    c=n//2
                    r = np.sqrt((yy-c)**2 + (xx-c)**2)
                    #classic donut
                    donut = (r >= 5) & (r <= 10)
                    
          
                    bg_donut = np.nanmean(apa[donut])
                    p2ll = peak
                    
                    CONDS.append(condition)
                    PHS.append(PHASE)
                    PHST.append(phase)
                    CLS.append(cluster)
                    SCORES.append(p2ll)
                    MARK.append(mark)
                    df1=pd.DataFrame([CONDS,PHS,PHST,CLS,SCORES,MARK],index=['condition','cell cycle phase','time point','cluster','APA score','mark']).T
                    DF.append(df1)

                 
                    #   plt.xlabel('relative position, kbp')
                 #   plt.ylabel('relative position, kbp')
                    title=f'{condition} {PHASE} {phase}_{mark} {cluster}'
                    plt.title(title)
                    plt.show()
```

### 3) Saddle Plot

```python
df_del=pd.read_csv('results_flow/delGAP1_8cl_rnds3.csv',index_col=0)
df_ran=pd.read_csv('results_flow/ranGAP1_8cl_rnds3.csv',index_col=0)
bins=pd.read_csv('comps/bins_comps_continuous.csv')


df_del['bin1_id']=[int(x.split('-')[0]) for x in df_del.index]
df_del['bin2_id']=[int(x.split('-')[1]) for x in df_del.index]
df_ran['bin1_id']=[int(x.split('-')[0]) for x in df_ran.index]
df_ran['bin2_id']=[int(x.split('-')[1]) for x in df_ran.index]
anno_del=cooler.annotate(df_del,bins[['deltaGAP1_G1s10h_continuous_rank']])
anno_ran=cooler.annotate(df_ran,bins[['ranGAP1_G1s10h_continuous_rank']])
```

```python
df=anno_del[(anno_del['deltaGAP1_G1s10h_continuous_rank1']>0)&(anno_del['deltaGAP1_G1s10h_continuous_rank2']>0)]
# columns
xcol = "deltaGAP1_G1s10h_continuous_rank1"
ycol = "deltaGAP1_G1s10h_continuous_rank2"
ccol = "cluster"



n_bins = 25
bins = np.linspace(0, 1, n_bins + 1)

def make_density_matrix(data, xcol, ycol, bins):
    counts, xedges, yedges = np.histogram2d(
        data[xcol],
        data[ycol],
        bins=[bins, bins]
    )

    density = counts / counts.sum()
    return density.T  


# background from all data
background = make_density_matrix(df, xcol, ycol, bins)

# avoid division by zero
eps = 1e-12
background = background + eps

for cl in sorted(df[ccol].unique()):

    sub = df[df[ccol] == cl]

    cluster_density = make_density_matrix(sub, xcol, ycol, bins)

    saddle = (cluster_density+1e-10) / (1e-10+background)
    saddle=np.log2(saddle)

    plt.figure(figsize=(6, 5))

    ax = sns.heatmap(
        saddle,
        cmap="coolwarm",
        center=0,
        vmin=-2,
        vmax=2,
        square=True,
    )

    # show axis values from 0 to 1
    tick_positions = np.linspace(0, n_bins, 6)
    tick_labels = np.round(np.linspace(0, 1, 6), 2)

    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels)

    ax.set_yticks(tick_positions)
    ax.set_yticklabels(tick_labels)

    plt.tight_layout()
```
